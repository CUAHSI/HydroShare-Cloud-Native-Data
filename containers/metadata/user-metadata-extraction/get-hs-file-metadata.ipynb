**Purpose**: Extract science metadata from a directory of files downloaded from HydroShare.

This notebook uses the `hsextract` docker container to perform metadata extraction.

```python
import os
from pathlib import Path
```

*NOTE: The HydroShare resource must be downloaded manually at this time*

```python
identifier = 'b1379f00121e456f958f9e22e913aa8a'
hsurl = f'https://hydroshare.org/resource/{identifier}/'
```

```python
# print the path to the content associated with the resource
data_path = Path(os.getcwd())/identifier/'data/contents'
print(data_path)
```

Pull the hsextract docker image. This will be used to extract science metadata from the files in the HydroShare resource.

```python
!docker pull scootna/hsextract:0.5.2
```

Mount the HydroShare data into the `hsextract` container and run the extraction procedure. The resulting output will be stored in a folder called `.hs`. 

```python
!docker run -v $data_path:/files scootna/hsextract:0.5.2 extract /files $hsurl
```

Move the extracted metadata into the `meta` folder because Jupyter doesn't enable us to view hidden directories within the lab interface.

```python
!mv $data_path/.hs $data_path/meta
```
