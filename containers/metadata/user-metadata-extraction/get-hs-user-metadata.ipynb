**Purpose**: Collect user defined metadata for content stored in HydroShare.

This uses code developed as part of the I-GUIDE Data Catalog to collect HydroShare data and coalese them using Pydantic models.

```python
import os
from pathlib import Path
from api.adapters import hydroshare as hsadapter
from api.models.catalog import DatasetMetadataDOC
```

```python
# Define the HydroShare resource that we're interested in.
identifier = '8e944afdd73d4cce945c3382bfcd25d9'
```

```python
# Query metadata from HydroShare
adapter = hsadapter.HydroshareMetadataAdapter()
metadata = await adapter.get_metadata(identifier)
```

```python
# preview the raw HydroShare metadata
metadata
```

```python
# get the metadata as pydantic classes
hs_metadata_model = hsadapter._HydroshareResourceMetadata(**metadata)
hs_metadata = hs_metadata_model.to_catalog_dataset()

# this should work, but returns None. I'm not sure why.
#
#hs_metadata = adapter.to_catalog_record(metadata)
```

```python
# preview to Schema.Org-aligned metadata
print(hs_metadata.json(indent=4))
```

```python
# write user metadata to file
metadata_path = Path(os.getcwd())/identifier/'data/contents/'
metadata_path.mkdir(parents=True, exist_ok=True)

user_metadata = metadata_path/"user_metadata.json"
with open(user_metadata, 'w') as f:
    f.write(hs_metadata.json(indent=4))
```
